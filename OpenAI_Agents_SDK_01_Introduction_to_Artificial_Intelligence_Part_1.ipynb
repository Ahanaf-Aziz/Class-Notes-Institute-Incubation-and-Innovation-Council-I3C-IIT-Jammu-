## 6. What Is Intelligence?

The word **intelligence** is common but difficult to define precisely.

A first-principles working definition is:

> **Intelligence is the capacity to use information effectively to achieve goals under varying conditions.**

This definition contains four central elements.

| Element | Meaning |
|---|---|
| **Information** | The system receives or possesses data about a situation. |
| **Processing** | It interprets, compares, transforms, or reasons about information. |
| **Goals** | It has a criterion for desirable outcomes. |
| **Adaptation** | It changes behaviour when circumstances change. |

Intelligence is not merely information storage. A dictionary contains information, but it does not independently use that information to pursue goals.

## 7. Intelligence as a Relationship

Intelligence is often best understood as a relationship among a **system**, a **task**, and an **environment**.

```text
                 ┌─────────────┐
                 │ Environment │
                 └──────┬──────┘
                        │ observations
                        ▼
┌──────────┐      ┌─────────────┐      ┌─────────┐
│  Goals   │ ───▶ │   System    │ ───▶ │ Actions │
└──────────┘      └─────────────┘      └────┬────┘
                                           │
                                           └── changes environment
```

A chess program may defeat expert players, but that does not mean it can prepare food, understand medical context, or comfort a distressed person.

> **Intelligence is normally evaluated relative to a domain, goal, and environment.**

### Discussion prompt

Is a calculator intelligent?

A calculator performs difficult operations quickly, yet it normally does not choose goals, adapt its strategy, or understand the significance of its result. Powerful performance is not automatically broad intelligence.

## 8. Characteristics of Human Intelligence

Human intelligence is not one ability. It is a family of interconnected capabilities.

| Capability | Description | Everyday Example |
|---|---|---|
| **Perception** | Interpreting sensory input | Recognising a person |
| **Attention** | Selecting relevant information | Focusing despite noise |

| **Social cognition** | Interpreting other people | Recognising sarcasm |

| **Adaptability** | Changing behaviour | Taking a new route |

## 9. What Makes Something Intelligent?

A system appears more intelligent when it demonstrates several of the following:

1. **Goal-directed behaviour** — actions contribute to an objective.
2. **Context sensitivity** — behaviour changes with the situation.
3. **Reasoning** — conclusions are derived from information.
4. **Adaptation** — behaviour changes through feedback.
5. **Generalisation** — prior knowledge transfers to new cases.
6. **Autonomy** — some actions occur without step-by-step control.
7. **Robustness** — performance continues under uncertainty or noise.
8. **Efficiency** — useful outcomes are reached without unnecessary cost.

### Complexity is not intelligence

A program may contain millions of lines of code yet remain unable to adapt. Another system may use a simple policy but respond effectively across changing situations.

### Behaviour and understanding

A machine may imitate intelligent behaviour without human-like consciousness or understanding. Engineering evaluation usually measures observable performance rather than inaccessible inner experience.

## 10. Human Intelligence vs Machine Intelligence

| Dimension | Human Intelligence | Machine Intelligence |
|---|---|---|
| **Learning** | Often learns from few examples and rich experience | Frequently needs data or designed feedback |
| **Speed** | Limited biological speed | Very fast for many computations |
| **Memory** | Associative, imperfect, contextual | Large and precise storage is possible |
| **Common sense** | Strong in familiar physical and social settings | Often fragile outside learned patterns |
| **Emotion** | Experiences and uses emotion | Can model emotional language without experience |
| **Generalisation** | Broad across everyday domains | Often uneven and domain-sensitive |
| **Consistency** | Affected by fatigue and bias | Repeatable, but errors can also be systematic |
| **Creativity** | Grounded in embodiment, culture, intention | Produces novel combinations from learned patterns |
| **Embodiment** | Operates through a living body | May be digital or connected to sensors |
| **Ethics** | Shaped by society and reflection | Must be designed, constrained, and governed |

The useful question is not merely “Which is more intelligent?” It is:

> **Which capabilities are stronger in humans, which are stronger in machines, and how should they work together?**

## 11. Narrow, General, and Superintelligence

### Narrow AI
Designed for a limited task or domain: spam filtering, recommendations, image classification, route planning, and many language-model applications.

### Artificial General Intelligence
A hypothetical or debated form of machine intelligence capable of flexible performance across a broad range of intellectual tasks.

### Artificial Superintelligence
A hypothetical intelligence exceeding human performance across most or all cognitively relevant domains.

```text
Narrow AI             General AI              Superintelligence
Specific tasks  ───▶  Broad capability  ───▶  Beyond human level
```

These categories are conceptual. Real systems may look broad while remaining unreliable in common sense, physical reasoning, unfamiliar contexts, or long-horizon planning.

## 13. Historical Definitions of AI

Four traditions are often used to organise definitions of AI.

| Orientation | Core Aim | Guiding Question |
|---|---|---|
| **Thinking humanly** | Model human thought | Does it think as humans think? |
| **Acting humanly** | Reproduce human-like behaviour | Does it behave like a human? |
| **Thinking rationally** | Apply valid reasoning principles | Does it reason correctly? |
| **Acting rationally** | Choose effective actions | Does it act toward goals? |

### Thinking humanly
Connected with cognitive science and psychology.

### Acting humanly
Focuses on observable behaviour; the Turing Test is historically linked with this view.

### Thinking rationally
Emphasises formal logic and valid inference.

### Acting rationally
Focuses on systems that select actions expected to achieve goals under constraints.

> A machine need not reproduce human mental processes exactly to produce useful intelligent behaviour.

## 14. AI as a Scientific Discipline

As science, AI investigates intelligence through models, theory, measurement, and experiments.

Researchers ask:

- How can knowledge be represented?
- How can systems learn?
- How does reasoning work under uncertainty?
- How do goals influence decisions?
- What enables generalisation?
- How should intelligent behaviour be measured?

```text
Question → Hypothesis → Model → Experiment → Observation → Revised Theory
```

Example scientific question:

> Does one planning algorithm solve a defined class of navigation problems more efficiently than another?

Answering it requires measurable definitions, controlled tasks, comparison, and analysis. AI science seeks both useful performance and explanatory understanding.

## 17. Traditional Programming vs Artificial Intelligence

Traditional programming commonly follows:

```text
Data + Explicit Rules → Output
```

Many machine-learning systems follow:

```text
Data + Desired Outputs → Learned Model
Learned Model + New Data → Prediction
```

| Dimension | Traditional Programming | AI / Machine Learning |
|---|---|---|
| Rules | Written by developers | Often learned from data |
| Best for | Stable, explicit logic | Complex patterns |
| Explainability | Often high | Varies |
| Adaptation | Requires code change | May improve through retraining |
| Data dependency | Usually moderate | Often substantial |
| Testing | Expected outputs | Metrics, robustness, drift |

### Important nuance

Rule-based systems can still be part of AI. Historical expert systems used large rule collections, and modern agents combine learned models with deterministic rules and guardrails.

## 18. Rule-Based Decision Making

A rule-based system represents knowledge with conditions and actions.

```text
IF condition is true
THEN perform an action
```

Rule-based systems are useful when:

- decision logic is explicit;
- domain rules are stable;
- accountability matters;
- outcomes must be predictable;
- and the number of cases is manageable.

They become difficult when inputs are ambiguous, exceptions multiply, rules interact, environments change, or learning is required.

The next examples increase gradually in structure. Each code cell is preceded by a conceptual explanation and followed by interpretation.

### Python Example 1 — Basic Temperature Classifier

This program receives one observation and maps it to a category. It is deterministic: the same input produces the same output.

Before running the cell, predict the result for `10`, `22`, and `35` degrees Celsius.

````python
def classify_temperature(temperature_celsius: float) -> str:
    """Classify temperature using explicit human-written rules."""
    if temperature_celsius < 18:
        return "Cold"
    elif temperature_celsius < 30:
        return "Comfortable"
    else:
        return "Hot"


test_temperatures = [10, 22, 35]

for temperature in test_temperatures:
    category = classify_temperature(temperature)
    print(f"{temperature:>2}°C → {category}")
````

#### Interpretation of Example 1

```text
Input
  │
  ▼
Temperature < 18?
  ├─ Yes → Cold
  └─ No
       ▼
Temperature < 30?
  ├─ Yes → Comfortable
  └─ No  → Hot
```

The program does not learn what “cold” means. The programmer selected the thresholds.

Limitations include climate differences, individual preferences, humidity, wind, clothing, and activity. The system cannot improve through experience.

> **A system can make decisions without learning.**

### Python Example 2 — Multi-Factor Rule-Based Advisor

A realistic decision can depend on several variables. This function uses temperature, rain, and air quality.

The order of rules matters: the safety condition is checked before convenience conditions.

````python
def recommend_activity(
    temperature_celsius: float,
    raining: bool,
    air_quality_index: int
) -> str:
    """Recommend an activity using prioritised rules."""

    if air_quality_index > 200:
        return "Stay indoors because air quality is unhealthy."

    if raining:
        return "Choose an indoor activity or carry rain protection."

    if 18 <= temperature_celsius <= 30:
        return "Outdoor activity is suitable."

    if temperature_celsius < 18:
        return "Outdoor activity is possible, but wear warm clothing."

    return "Limit strenuous outdoor activity because it is very hot."


scenarios = [
    {"temperature_celsius": 24, "raining": False, "air_quality_index": 75},
    {"temperature_celsius": 24, "raining": True, "air_quality_index": 75},
    {"temperature_celsius": 32, "raining": False, "air_quality_index": 250},
]

for number, scenario in enumerate(scenarios, start=1):
    print(f"Scenario {number}: {scenario}")
    print(recommend_activity(**scenario))
    print()
````

#### Interpretation of Example 2

The system is context-sensitive because several features influence its response. Yet all behaviour remains human-authored.

In the third scenario, unhealthy air quality overrides temperature because the corresponding rule is evaluated first.

This introduces a **decision policy**:

> A decision policy maps situations to actions.

Later agent policies may incorporate model reasoning, tool selection, safety constraints, confidence thresholds, human approval, and recovery behaviour.

### Python Example 3 — Separating Rules from the Decision Engine

Long conditional blocks become difficult to maintain. The next design stores rules separately from the engine.

Each rule has:

- a name;
- a condition;
- an action.

The engine evaluates rules in priority order and reports which rule matched, improving traceability.

````python
from dataclasses import dataclass
from typing import Callable


@dataclass
class StudentCase:
    attendance_percentage: float
    average_score: float
    assignments_pending: int


@dataclass
class Rule:
    name: str
    condition: Callable[[StudentCase], bool]
    action: str


rules = [
    Rule(
        name="Critical attendance rule",
        condition=lambda case: case.attendance_percentage < 60,
        action="Schedule an academic counselling meeting."
    ),
    Rule(
        name="Pending work rule",
        condition=lambda case: case.assignments_pending >= 3,
        action="Create an assignment completion plan."
    ),
    Rule(
        name="Strong performance rule",
        condition=lambda case: (
            case.average_score >= 85
            and case.assignments_pending == 0
        ),
        action="Recommend an advanced enrichment task."
    ),
]


def evaluate_student(case: StudentCase) -> dict:
    """Return the first matching rule and action."""
    for rule in rules:
        if rule.condition(case):
            return {
                "matched_rule": rule.name,
                "recommended_action": rule.action
            }

    return {
        "matched_rule": "Default rule",
        "recommended_action": "Continue regular monitoring."
    }


student = StudentCase(
    attendance_percentage=78,
    average_score=72,
    assignments_pending=4
)

result = evaluate_student(student)

print("Student case:", student)
print("Matched rule:", result["matched_rule"])
print("Recommended action:", result["recommended_action"])
````

#### Interpretation of Example 3

```text
Student Data → Rule Collection → Decision Engine → Rule + Action
```

Benefits:

- rules can be inspected independently;
- every decision identifies the matched rule;
- priorities can be changed by reordering;
- rules can be added without rewriting the engine.

Limitations remain. The system cannot discover new rules, understand ambiguous circumstances, weigh several partially relevant rules, or adapt automatically.

Modern agents often combine model-based interpretation with deterministic rules for policy and safety.
