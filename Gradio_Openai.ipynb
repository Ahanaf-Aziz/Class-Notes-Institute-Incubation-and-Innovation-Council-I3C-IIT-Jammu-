```python
from openai import OpenAI

openai = OpenAI(api_key = os.environ["OPENAI_API_KEY"])
```

```python
#importing gradio

import gradio as gr
```

```python
def greet(name):
  return 'Hello ' + name + " !."

# gr.Interface(
#     fn=greet,
#     inputs="text",
#     outputs="text"
# ).launch()

demo = gr.Interface(
    fn=greet,
    inputs="text",
    outputs="text"
)

demo.launch()
```

```python
#Multiple inputs

def add(a, b):
  return a + b

demo = gr.Interface(
    fn=add,
    inputs = [
        gr.Number(label = "First Number"),
        gr.Number(label = "Second Number")
    ],
    outputs="number"
)

demo.launch()
```

```python
#Using Sliders

def powered(number, exponent):
  return number ** exponent

gr.Interface(
    fn=powered,
    inputs = [
        gr.Number(label = "Enter the number"),
        gr.Slider(minimum = 0, maximum = 20, label = "Exponent")
    ],
    outputs = "number"
).launch()
```

```python
#Using Images

from PIL import Image

def convert_grayscale(image):
  return image.convert('L')

demo = gr.Interface(
    fn=convert_grayscale,
    inputs = gr.Image(type="pil"),
    outputs = gr.Image()
)

demo.launch()
```

```python
#Advanced Blocks

with gr.Blocks() as demo:

  gr.Markdown("# Text Reverser App")

  input_text = gr.Textbox(label="Enter Text")
  output_text = gr.Textbox(label = "Reversed Text")

  button = gr.Button("Reverse")

  def reverse_text(text):
    return text[::-1]

  button.click(
      reverse_text,
      inputs = input_text,
      outputs = output_text,
  )

demo.launch()
```

```python
def analyse_text(text):

  words=len(text.split())
  characters = len(text)
  sentences = text.count(".")

  result = f"""

  Text Ananlysis Results:

  Words: {words}
  Characters: {characters}
  Sentences: {sentences}
  """

  return result

demo = gr.Interface(
    fn = analyse_text,
    inputs = gr.Textbox(lines=5, placeholder="Enter your text here.."),
    outputs = "text",
    title = "AI Text Analyzer",
    description = "Analyse text statistics instantly.",
)

demo.launch(debug=True)
```

```python
# WITH OPENAI

def askgpt(question):

  response = openai.chat.completions.create(
      model="gpt-4o-mini",
      messages= [
          {"role": "user", "content": question}
      ]
  )

  return response.choices[0].message.content
```

```python
demo = gr.Interface(
    fn = askgpt,
    inputs = gr.Textbox(lines=2, placeholder="Ask a Question"),
    outputs = "text",
    title = "Ask GPT",
    description = "Get answers from Chatgpt"
)

demo.launch()
```

```python
#Chatgpt style Chatbot with Gradio

def chatbot(message, history):

  messages = []

  for user, bot in history:
    message.append({"role": "user", "content": user})
    message.append({"role": "assistant", "content": bot})

  messages.append({"role": "user", "content": message})

  response = openai.chat.completions.create(
      model="gpt-4o-mini",
      messages = messages
  )

  return response.choices[0].message.content

gr.ChatInterface(
    fn=chatbot,
    ).launch()
```
