## 3. From Programs to Agents

A normal program may simply transform an input into an output.

```text
Input → Program → Output
```

An agent participates in a continuing interaction:

```text
Environment → Observation → Agent → Action → Environment
```

The environment may change after the action, producing a new observation.

### Example

A currency converter receives an amount and returns a converted value. It does not usually select goals or continue acting.

A travel-booking agent may:

- receive the user's destination and dates;
- search multiple services;
- compare options;
- ask for missing information;
- request approval;
- book a selected itinerary;
- record the result;
- and recover from errors.

The second system is agent-like because it performs a sequence of context-dependent actions toward a goal.

## 4. What Is an Agent?

A useful first-principles definition is:

> **An agent is a system that perceives an environment through observations and acts upon that environment in pursuit of goals.**

This definition contains four essential parts:

| Component | Meaning |
|---|---|
| **Agent** | The decision-making system |
| **Environment** | Everything relevant outside the agent |
| **Observation** | Information received by the agent |
| **Action** | Something the agent does that may affect the environment |

A more detailed view is:

```text
                     observations
Environment  ─────────────────────────▶  Agent
     ▲                                      │
     │                                      │ actions
     └──────────────────────────────────────┘
```

The agent may also contain:

- internal state;
- memory;
- rules;
- a model;
- tools;
- plans;
- goals;
- and safety constraints.

## 5. Agent, Model, and Application

These terms are related but should not be confused.

| Term | Meaning |
|---|---|
| **Model** | A computational component that transforms inputs into outputs |
| **Agent** | A system that selects and performs actions toward goals |
| **Application** | The broader product containing user interfaces, agents, databases, policies, and services |

A language model may be used inside an agent.

```text
Application
├── User interface
├── Agent
│   ├── Language model
│   ├── Tools
│   ├── Memory
│   ├── Rules
│   └── Control loop
├── Database
└── Monitoring
```

A model can answer a question without being a complete agent.  
An agent uses models and other components to decide what to do next.

## 6. The Agent–Environment Loop

The agent loop can be represented as:

```text
1. Observe
2. Interpret
3. Decide
4. Act
5. Receive feedback
6. Repeat
```

Expanded:

```text
┌─────────────┐
│ Environment │
└──────┬──────┘
       │ observation
       ▼
┌─────────────┐
│    Agent    │
│ perception  │
│ state       │
│ decision    │
└──────┬──────┘
       │ action
       ▼
┌─────────────┐
│ Environment │
└─────────────┘
```

This loop may run:

- once;
- for a fixed number of steps;
- until a goal is achieved;
- until a stopping condition occurs;
- or continuously.

## 8. Observations, Percepts, and Percept History

An individual piece of information received by an agent is often called a **percept** or observation.

Examples:

- “The user asked for a refund.”
- “The API returned status 404.”
- “The room temperature is 31°C.”
- “The account balance is £120.”
- “The previous tool call failed.”

A single observation may be insufficient. Agents often need the history of observations.

```text
Observation 1 → Observation 2 → Observation 3 → Current Decision
```

This sequence is called the **percept history**.

### Why history matters

A support agent should not repeat a question that the user already answered.  
A navigation agent should remember previously blocked routes.  
A workflow agent should know which tasks have already completed.

## 9. Goals and Performance Measures

A goal describes a desired state or outcome.

Examples:

- deliver a package;
- answer a user accurately;
- minimise travel time;
- resolve a support ticket;
- produce a valid report;
- avoid unsafe actions.

However, goals can be vague. A **performance measure** makes success more operational.

| Goal | Possible Performance Measures |
|---|---|
| Answer questions | accuracy, citation quality, response time |
| Plan travel | price, duration, number of stops |
| Detect fraud | recall, precision, false positive rate |
| Support students | resolution rate, satisfaction, escalation accuracy |

A strong agent design distinguishes:

```text
Goal = What should be achieved?
Performance measure = How will success be judged?
```

## 10. Why Performance Measures Matter

Consider the goal:

> “Provide fast customer support.”

An agent optimised only for speed may give incorrect answers.

A better performance measure may combine several factors:

```text
Overall performance
= accuracy
+ response speed
+ policy compliance
+ user satisfaction
- unsafe actions
```

This creates a trade-off.

### Multi-objective design

Real systems usually optimise several objectives:

- quality;
- speed;
- cost;
- safety;
- privacy;
- user effort;
- and reliability.

Poorly defined performance measures can produce undesirable behaviour even when the agent technically follows its objective.

## 11. Rational Behaviour

A rational agent selects an action expected to perform well according to its performance measure, given:

- its observations;
- its prior knowledge;
- its available actions;
- and its computational limits.

A simplified definition is:

> **A rational action is the action expected to produce the best available outcome under current information.**

Rational does not mean perfect.

### Example

A route-planning agent selects the fastest visible route. An unexpected road closure later causes delay.

The decision may still have been rational if the closure was unknowable at decision time.

## 12. Bounded Rationality

Real agents have limits:

- incomplete information;
- limited time;
- limited memory;
- limited computation;
- imperfect models;
- and uncertain environments.

This gives rise to **bounded rationality**.

> **Bounded rationality means selecting a sufficiently good action under practical constraints rather than calculating a perfect action.**

Examples:

- a chess agent searches only a limited number of future moves;
- a support agent uses a time limit before escalating;
- a shopping agent examines ten relevant products rather than every product;
- a scheduling agent chooses a workable slot without evaluating every possible calendar arrangement.

Production systems must explicitly manage these constraints.

## 13. Rationality Is Not Omniscience

Three concepts must be separated.

| Concept | Meaning |
|---|---|
| **Rationality** | Choosing well using available information |
| **Omniscience** | Knowing the actual outcome of every action |
| **Success** | The outcome happens to be favourable |

A rational agent can fail because:

- the world changes;
- observations are incomplete;
- tools return incorrect data;
- the model makes an error;
- or random events occur.

Therefore, evaluation should consider both decision quality and outcome quality.

## 14. Autonomy vs Automation

These terms are often confused.

### Automation

Automation executes predefined processes with limited variation.

```text
Trigger → Fixed Workflow → Output
```

Example: send a welcome email after registration.

### Autonomy

Autonomy means the system has some freedom to select actions based on context.

```text
Goal + Context → Agent Chooses Action
```

Example: inspect a support request, decide whether to answer, search, ask a question, or escalate.

| Dimension | Automation | Autonomy |
|---|---|---|
| Flow | Predefined | Context-dependent |
| Decision freedom | Low | Higher |
| Predictability | High | Lower |
| Adaptability | Limited | Greater |
| Risk | Usually narrower | Requires stronger controls |

An agent can be partially autonomous rather than fully autonomous.

## 15. Levels of Agent Autonomy

Autonomy can be viewed as a spectrum.

```text
Level 0: Human performs every action
Level 1: System suggests
Level 2: System acts after approval
Level 3: System acts within limits
Level 4: System acts and reports
Level 5: System operates independently
```

Examples:

| Level | Example |
|---|---|
| Suggest | Draft an email for review |
| Approval required | Prepare a refund, then request confirmation |
| Bounded action | Reschedule only within approved working hours |
| Act and report | Archive low-priority notifications and send a summary |
| Independent | Rare and suitable only for low-risk, tightly constrained domains |

> **Design principle**
>
> Increase autonomy only when evaluation, monitoring, safeguards, and recovery mechanisms are strong enough.

## 16. Types of Environments

Agent environments can be classified along several dimensions.

| Dimension | Type A | Type B |
|---|---|---|
| Observability | Fully observable | Partially observable |
| Determinism | Deterministic | Stochastic |
| Time structure | Episodic | Sequential |
| Change | Static | Dynamic |
| Representation | Discrete | Continuous |
| Participants | Single-agent | Multi-agent |
| Knowledge | Known | Unknown |

These dimensions affect agent design.

A fully observable, deterministic environment is easier than a partially observable, stochastic, dynamic environment.

## 17. Fully vs Partially Observable

### Fully observable

The agent can access all information relevant to the current decision.

Example: a board game where the full board is visible.

### Partially observable

Some relevant information is hidden, delayed, noisy, or unavailable.

Examples:

- medical diagnosis;
- financial markets;
- customer intent;
- real-world driving;
- distributed software systems.

In partially observable environments, agents may require:

- memory;
- inference;
- uncertainty estimates;
- clarification questions;
- or external tools.

## 18. Deterministic vs Stochastic

### Deterministic environment

A given action in a given state always produces the same outcome.

```text
State + Action → Predictable Next State
```

### Stochastic environment

Outcomes involve uncertainty or randomness.

```text
State + Action → Distribution of Possible Outcomes
```

Examples:

| Deterministic | Stochastic |
|---|---|
| simple calculator | stock market |
| fixed rule engine | weather |
| puzzle with fixed transitions | medical treatment response |
| controlled simulation | human conversation |

Most real-world agent environments contain at least some uncertainty.

## 19. Episodic vs Sequential

### Episodic

Each decision is largely independent.

Example: classify each incoming image separately.

### Sequential

Current actions affect future states and options.

Examples:

- travel planning;
- negotiation;
- game playing;
- research workflows;
- multi-step customer support.

```text
Action at time t
      │
      ▼
Changes state at time t+1
      │
      ▼
Changes future decisions
```

Agent systems are often sequential, which makes memory and planning important.

## 20. Static vs Dynamic

### Static environment

The environment does not change while the agent is deciding.

### Dynamic environment

The environment may change during decision-making.

Examples:

- a saved document is relatively static;
- traffic is dynamic;
- a live auction is dynamic;
- a shared calendar is dynamic;
- inventory is dynamic.

Dynamic environments require:

- fresh observations;
- time-sensitive actions;
- concurrency handling;
- and re-planning.

## 21. Agent Architectures

An agent architecture is the internal structure used to map observations to actions.

This notebook introduces four major forms:

1. **Simple reflex agent**
2. **Model-based agent**
3. **Goal-based agent**
4. **Utility-based agent**

```text
Reflex → Model-Based → Goal-Based → Utility-Based
Simple                                     Richer decision-making
```

These are conceptual categories. Real agents often combine features from several architectures.

## 22. Simple Reflex Agents

A simple reflex agent selects an action using the current observation only.

```text
Current Observation → Condition–Action Rule → Action
```

Example:

```text
IF email contains "urgent"
THEN assign high priority
```

Strengths:

- simple;
- fast;
- predictable;
- easy to test.

Limitations:

- no memory;
- no understanding of hidden state;
- weak in partially observable environments;
- brittle when conditions change.

### Python Example 1 — Simple Reflex Agent

The agent below controls a basic room fan. It reads only the current temperature and selects an action.

Before running the code, predict the output for `20`, `27`, and `34`.

````python
def fan_agent(temperature_celsius: float) -> str:
    """A simple reflex agent using only the current observation."""
    if temperature_celsius >= 32:
        return "Set fan to HIGH"
    elif temperature_celsius >= 25:
        return "Set fan to MEDIUM"
    else:
        return "Set fan to LOW"


for temperature in [20, 27, 34]:
    action = fan_agent(temperature)
    print(f"Observation: {temperature}°C → Action: {action}")
````

#### Interpretation of Example 1

The agent uses only the current observation.

```text
Temperature → Rule Matching → Fan Action
```

It does not know:

- how long the room has been hot;
- whether the fan was already running;
- whether anyone is present;
- or how quickly the temperature is changing.

This is appropriate for a simple, fully observable task but weak for richer environments.

## 23. Model-Based Agents

A model-based agent maintains internal state about the world.

```text
Previous State
      +
Current Observation
      ↓
Updated Internal State
      ↓
Action
```

Internal state is useful when the environment is partially observable.

Example:

A cleaning robot may not currently see every room, but it can remember which rooms were already cleaned.

The “model” here means a representation of how the world changes. It does not necessarily mean a machine-learning model.

### Python Example 2 — Model-Based Cleaning Agent

This agent remembers which rooms have already been cleaned. The current observation alone is not enough; internal state influences the action.

````python
class CleaningAgent:
    """A small model-based agent with internal memory."""

    def __init__(self):
        self.cleaned_rooms = set()

    def act(self, room: str, is_dirty: bool) -> str:
        if is_dirty:
            self.cleaned_rooms.add(room)
            return f"Clean {room}"

        if room not in self.cleaned_rooms:
            self.cleaned_rooms.add(room)
            return f"Mark {room} as checked"

        return f"Move to another room; {room} is already handled"


agent = CleaningAgent()

observations = [
    ("Room A", True),
    ("Room B", False),
    ("Room A", False),
]

for room, is_dirty in observations:
    action = agent.act(room, is_dirty)
    print(f"Observation: room={room}, dirty={is_dirty}")
    print(f"Action: {action}")
    print(f"Memory: {agent.cleaned_rooms}
")
````

#### Interpretation of Example 2

The agent maintains a set of handled rooms.

Its action depends on both:

```text
Current Observation + Stored State
```

This prevents repeated work.

In future SDK-based agents, state may include:

- conversation history;
- user preferences;
- completed tool calls;
- previous failures;
- structured session data;
- and task progress.

## 24. Goal-Based Agents

A goal-based agent evaluates actions according to whether they help achieve a target state.

```text
Current State + Goal + Possible Actions → Selected Action
```

Example:

A travel agent with the goal “reach the airport by 18:00” may compare:

- taxi;
- bus;
- train;
- and walking.

The current situation alone does not determine the action. The goal gives direction.

Goal-based behaviour often requires search or planning.

### Python Example 3 — Goal-Based Route Selection

The following agent selects routes that satisfy a maximum travel time and then chooses the shortest valid route.

The program illustrates how a goal constrains acceptable actions.

````python
routes = [
    {"name": "Route A", "minutes": 45, "available": True},
    {"name": "Route B", "minutes": 30, "available": False},
    {"name": "Route C", "minutes": 38, "available": True},
]


def choose_route(route_options, maximum_minutes):
    """Choose the fastest available route that satisfies the goal."""
    valid_routes = [
        route for route in route_options
        if route["available"] and route["minutes"] <= maximum_minutes
    ]

    if not valid_routes:
        return None

    return min(valid_routes, key=lambda route: route["minutes"])


goal_limit = 40
selected = choose_route(routes, goal_limit)

print(f"Goal: arrive within {goal_limit} minutes")
print("Selected route:", selected)
````

#### Interpretation of Example 3

The agent:

1. represents the goal as a maximum travel time;
2. filters unavailable or unacceptable actions;
3. selects the best remaining route.

This is more flexible than a fixed rule such as “always choose Route A.”

However, the agent still considers only one criterion. Real route decisions may include cost, safety, reliability, and user preference.

## 25. Utility-Based Agents

A utility-based agent assigns a score to outcomes and chooses the action with the highest expected utility.

```text
Possible Action → Expected Outcome → Utility Score
```

Utility is useful when several goals conflict.

Example:

A route may be:

- fast but expensive;
- cheap but slow;
- comfortable but unreliable.

A utility function converts these factors into a comparable score.

> **Important**
>
> Utility functions encode values and priorities. Poorly designed utility can produce poor behaviour.

### Python Example 4 — Utility-Based Route Agent

The following agent balances travel time and cost. Lower time and lower cost receive better scores.

The weights represent priorities and can be changed.

````python
route_options = [
    {"name": "Taxi", "minutes": 25, "cost_gbp": 24},
    {"name": "Train", "minutes": 35, "cost_gbp": 8},
    {"name": "Bus", "minutes": 50, "cost_gbp": 4},
]


def route_utility(route, time_weight=0.6, cost_weight=0.4):
    """Higher utility is better."""
    time_score = 100 - route["minutes"]
    cost_score = 100 - route["cost_gbp"]
    return time_weight * time_score + cost_weight * cost_score


for route in route_options:
    route["utility"] = route_utility(route)

best_route = max(route_options, key=lambda route: route["utility"])

for route in route_options:
    print(
        f'{route["name"]}: '
        f'time={route["minutes"]} min, '
        f'cost=£{route["cost_gbp"]}, '
        f'utility={route["utility"]:.1f}'
    )

print("
Selected route:", best_route["name"])
````

#### Interpretation of Example 4

The agent combines two preferences:

- speed;
- cost.

Changing the weights changes behaviour.

For a traveller who values time strongly, increase `time_weight`.  
For a traveller who prioritises budget, increase `cost_weight`.

This demonstrates that agent behaviour depends not only on intelligence but also on **what the system is optimised to value**.

## 26. Reactive vs Deliberative Behaviour

### Reactive agent

Responds quickly to current conditions.

```text
Observation → Immediate Action
```

Examples:

- collision avoidance;
- spam blocking;
- emergency shutdown.

### Deliberative agent

Builds plans or compares future actions.

```text
State → Possible Futures → Plan → Action
```

Examples:

- travel planning;
- project scheduling;
- research strategy;
- supply-chain optimisation.

### Hybrid agent

Most practical systems combine both:

```text
Reactive safety layer
        +
Deliberative planning layer
```

A self-driving system may plan a route deliberatively while reacting immediately to an obstacle.

## 27. Agent Failure Modes

Even simple agents can fail in predictable ways.

| Failure Mode | Description |
|---|---|
| Poor observations | Inputs are missing, stale, or incorrect |
| Wrong goal | The agent optimises the wrong outcome |
| Bad performance measure | Metrics encourage undesirable behaviour |
| Inadequate state | Important history is forgotten |
| Tool failure | APIs or services return errors |
| Over-autonomy | The agent acts without suitable approval |
| Rule conflict | Multiple policies disagree |
| Infinite loop | No correct stopping condition |
| Unsafe action | Constraints are missing |
| No recovery | Failure is detected but not handled |

Reliable agent engineering begins by anticipating these failures.
