# Text Cleaning

Text cleaning is the process of **removing noise from raw text** so that machines can process it effectively.

## Why is Text Cleaning Important?

Raw text often contains:

- Uppercase and lowercase variations
- Punctuation and symbols
- Numbers and emojis
- Extra white spaces

These add **noise**, not meaning.

## Common Text Cleaning Operations

- Convert text to lowercase
- Remove punctuation
- Remove numbers
- Remove special characters
- Normalize white spaces

## Benefits of Text Cleaning

- Smaller vocabulary
- Better model performance
- Faster processing

````python
#1. Text lowercasing

text1 = "Hello World"
print(text1.lower())

text2 = "NLP is FuN"
print(text2.lower())
````

````python
#2. Removing Punctuations

import string

text1 = "Hello! How are you?"
print(text1.translate(str.maketrans("", "", string.punctuation)))

text2 = "I'm enjoying NLP."
print(text2.translate(str.maketrans("", "", string.punctuation)))
````

````python
#3. Remove Numbers

import re  #regex library

text1 = "NLP 101 is easy"
print(re.sub(r'\d+', '', text1)) #remove digits

text2 = "I have 3 dogs and 2 cats"
print(re.sub(r'\d+', '', text2))

text = "Hello, World! How's it going? #NLP 😊" #capture punctuation
punct = re.findall(r"[^\w\s]", text)
print(punct)
````

##4. Tokenization

Tokenization is the process of **breaking text into smaller units** called **tokens**.

### Types of Tokenization

1. **Word Tokenization** 📝  
   Splits text into individual words.

2. **Sentence Tokenization** 📄  
   Splits text into individual sentences.

### Why is Tokenization Important?

- Models do not understand raw text.
- Tokens are the basic units used in NLP.
- Most NLP operations work on tokens, not full text.

### Example

**Input:**

```text
I love NLP
```

**Output:**

```python
["I", "love", "NLP"]
```

````python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt_tab")  # download tokenizer model

text1 = "NLP is amazing"
print(word_tokenize(text1))

text2 = "I have a cat and a dog"
print(word_tokenize(text2))
````

## 5. Stopword Removal

Stopwords are **very common words** that usually do not add much meaning to a sentence.

### Examples of Stopwords

- is
- the
- and
- of
- to
- in

### Why Remove Stopwords?

- They appear in almost every document.
- They increase the feature vector size.
- They dilute important information.

### Benefits of Stopword Removal

- Text becomes more meaningful.
- Feature space is reduced.
- Improves the efficiency of many NLP models.

````python
from nltk.corpus import stopwords

nltk.download("stopwords")

# Load English stopwords
stop_words = set(stopwords.words("english"))

tokens1 = word_tokenize("NLP is a subfield of AI")
print(tokens1)
print([w for w in tokens1 if w.lower() not in stop_words])

tokens2 = word_tokenize("I am a student")
print(tokens2)
print([w for w in tokens2 if w.lower() not in stop_words])
````

# Stemming

Stemming reduces words to their **root form** using **rule-based methods**.

## Examples

- `running` → `run`
- `studies` → `studi`
- `playing` → `play`

## Characteristics of Stemming

- Fast
- Rule-based
- Output may not be a real dictionary word

## Use Cases

- Search engines
- Information retrieval
- Text classification
- Spam detection

````python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words1 = ["running", "runs", "ran"]
print([stemmer.stem(word) for word in words1])

words2 = ["study", "studies", "studying"]
print([stemmer.stem(word) for word in words2])
````

# Lemmatization

Lemmatization reduces words to their **dictionary base form (lemma)**.

## Examples

- `running` → `run`
- `studies` → `study`
- `better` → `good`

## Characteristics of Lemmatization

- Uses vocabulary and grammar
- More accurate than stemming
- Slower than stemming

## Preferred In

- Chatbots
- NLP pipelines
- Semantic analysis
- Machine translation


````python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')  #dictionary for lemmatization
lemmatizer = WordNetLemmatizer() #initialize

words1 = ["running", "runs", "ran"]
print([lemmatizer.lemmatize(word) for word in words1])
````

# One-Hot Encoding

One-Hot Encoding represents each word as a **binary vector**.

## How It Works

- Vocabulary size = Vector length
- Each word gets a unique index
- Presence → `1`
- Absence → `0`

## Example Vocabulary

```python
["ai", "love", "nlp"]
```

**Encoding:**

```python
"nlp"  → [0, 0, 1]
"love" → [0, 1, 0]
"ai"   → [1, 0, 0]
```

## Limitations

- Creates high-dimensional vectors for large vocabularies.
- Does not capture semantic meaning between words.
- Sparse representation (mostly zeros).
- Ignores the order of words.

````python
import numpy as np

vocab = ["ai", "data", "learning", "nlp"]

def one_hot(word, vocab):
    vector = np.zeros(len(vocab))
    print (vector)

    if word in vocab:
      vector[vocab.index(word)] = 1
    return vector


for word in vocab:
  print(word,"->", one_hot(word, vocab))
````

````python
from sklearn.feature_extraction.text import CountVectorizer

sentences = ["I love NLP", "AI is fun"]

vectorizer = CountVectorizer(binary=True)   # Create a CountVectorizer object

X = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(X)
print(X.toarray())    #One-hot vectors
````

# ⚠️ Important: Core Difference Between the Two Outputs

Although both outputs look like **one-hot encoding**, they represent **different things**.

## 1. Word-Level One-Hot Encoding

- Each **individual word** is represented by its own vector.
- The vector has **exactly one `1`** and the remaining values are `0`.
- Used to uniquely identify a word.

**Example:**

```python
Vocabulary = ["ai", "love", "machine", "nlp"]

"nlp" → [0, 0, 0, 1]
```

---

## 2. Document-Level Binary Encoding (CountVectorizer)

- Each **document (sentence)** is represented by a vector.
- The vector indicates the **presence (`1`) or absence (`0`)** of words in the vocabulary.
- A document can contain **multiple `1`s** because it may contain multiple words.
- This is actually a **Binary Bag of Words**, not true one-hot encoding.

### Example

**Vocabulary:**

```python
["ai", "love", "machine", "nlp"]
```

**Sentence:**

```text
I love NLP
```

**Binary Bag of Words:**

```python
[0, 1, 0, 1]
```

---

## In Short

| Word-Level One-Hot Encoding | Document-Level Binary Encoding |
|-----------------------------|--------------------------------|
| Encodes **individual words** | Encodes **entire documents/sentences** |
| Exactly **one `1`** in every vector | Can contain **multiple `1`s** |
| Identifies a single word | Indicates which words appear in a document |
| Used for word representation | Used for document representation (e.g., `CountVectorizer(binary=True)`) |

# Bag of Words

Bag of Words represents text by **counting the frequency of words** in a document.

## Key Ideas

- Ignores grammar and word order.
- Considers how many times each word appears.
- Creates a **document-word matrix**.
- Each row represents a document, and each column represents a word in the vocabulary.

## Example

**Sentence:**

```text
I love love NLP
```

**Vocabulary:**

```python
["I", "love", "NLP"]
```

**Bag of Words Vector:**

```python
[1, 2, 1]
```

Here, `"love"` appears **2** times.

---

## Advantages

- Simple and easy to understand.
- Easy to implement.
- Works well as a baseline for text classification tasks.

## Limitations

- Ignores word order and grammar.
- Does not capture the meaning or context of words.
- Common words can dominate the representation.
- Produces high-dimensional and sparse vectors for large vocabularies.

````python
vectorizer = CountVectorizer()  #Counts word frequency

sentences = ["I love NLP. I also love pizza.","I am typing and I love AI"]
X = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(X.toarray())
````

````python
# Import the required class
from sklearn.feature_extraction.text import CountVectorizer

# Input documents
documents = [
    "I love love NLP",
    "I love AI",
    "NLP and AI are powerful"
]

# Initialize CountVectorizer with stopword removal
vectorizer = CountVectorizer(stop_words='english')

# Fit the vectorizer and transform documents into BoW matrix
bow_matrix = vectorizer.fit_transform(documents)

# Display the learned vocabulary
print("Vocabulary (column order):")
print(vectorizer.get_feature_names_out())

# Display the Bag of Words matrix
print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
````

# TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF measures **how important a word is to a document** relative to a collection of documents.

## Intuition

- Words that appear frequently in one document but are rare across all documents are considered important.
- Common words receive lower weights.

## Components

- **TF (Term Frequency):** How often a word appears in a document.
- **IDF (Inverse Document Frequency):** How rare the word is across all documents.

## Why TF-IDF is Better than Bag of Words (BoW)

- Reduces the impact of very common words.
- Highlights more meaningful and informative terms.
- Produces better features for many NLP tasks like text classification and information retrieval.

````python
# Import TF-IDF Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Input documents (same as previous examples)
documents = [
    "I love love NLP",
    "I love AI",
    "NLP and AI are powerful"
]

# Initialize TF-IDF vectorizer with stopword removal
tfidf_vectorizer = TfidfVectorizer(stop_words='english')

# Fit the model and transform documents into TF-IDF matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Display the learned vocabulary (column order)
print("Vocabulary (column order):")
print(tfidf_vectorizer.get_feature_names_out())

# Display the TF-IDF matrix
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
````

# Part-of-Speech (POS) Tagging

POS Tagging assigns **grammatical roles** to words in a sentence.

## Common POS Tags

- Noun
- Verb
- Adjective
- Adverb
- Pronoun
- Preposition

## Example

**Sentence:**

```text
I love NLP
```

**POS Tags:**

- **I** → Pronoun
- **love** → Verb
- **NLP** → Proper Noun

## Applications

- Grammar checking
- Keyword extraction
- Syntax analysis
- Named Entity Recognition (NER)
- Machine Translation

````python
import nltk
from nltk import pos_tag
from nltk.tokenize import word_tokenize

# Download the POS tagger model
nltk.download('averaged_perceptron_tagger_eng')

# Tokenize the sentence
tokens = word_tokenize("I love NLP")

# Perform POS tagging
print(pos_tag(tokens))
````
